```python
# ==============================================================================
# TRACE THE ACE — OOF BLEND + CALIBRATION
# CELL 1 — OOF ARTIFACT DISCOVERY + KEYED ALIGNMENT LOCK — REPAIR v2
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 100)
print("TRACE THE ACE — OOF BLEND + CALIBRATION")
print("CELL 1 — OOF ARTIFACT DISCOVERY + KEYED ALIGNMENT LOCK")
print("=" * 100)


# ==============================================================================
# 0. DEPENDENCY GATE
# ==============================================================================

assert "PROJECT_ROOT" in globals(), (
    "Cell 0 dependency failed. Run Cell 0 first."
)

assert "SCRATCH_ROOT" in globals(), (
    "Cell 0 dependency failed."
)

assert "LOCKED_BLEND_WEIGHTS" in globals(), (
    "Cell 0 dependency failed: blend weights missing."
)

print("\nCell 0 dependency : PASS")


# ==============================================================================
# 1. DISCOVER OOF ARTIFACTS
# ==============================================================================

PROJECT_ROOT = Path(PROJECT_ROOT).resolve()


def discover_unique_artifact(
    filename,
    preferred_tokens=None,
):
    candidates = sorted(
        PROJECT_ROOT.rglob(filename)
    )

    assert len(candidates) > 0, (
        f"Artifact not found: {filename}"
    )

    if preferred_tokens:
        preferred = [
            p
            for p in candidates
            if all(
                token.lower() in str(p).lower()
                for token in preferred_tokens
            )
        ]

        if len(preferred) == 1:
            return preferred[0]

        if len(preferred) > 1:
            candidates = preferred

    # Deterministic newest-artifact selection.
    return max(
        candidates,
        key=lambda p: p.stat().st_mtime,
    )


MODERNBERT_OOF_PATH = discover_unique_artifact(
    "modernbert_5fold_oof.parquet",
    preferred_tokens=["oof"],
)

STRUCTURED_OOF_PATH = discover_unique_artifact(
    "structured_prior_oof.parquet",
    preferred_tokens=["09C", "structured_prior"],
)

TFIDF_OOF_PATH = discover_unique_artifact(
    "tfidf_5fold_oof.parquet",
    preferred_tokens=["09C", "tfidf"],
)


print("\n" + "=" * 100)
print("LOCKED OOF ARTIFACTS")
print("=" * 100)

print(
    f"ModernBERT       : {MODERNBERT_OOF_PATH}"
)

print(
    f"Structured+Prior : {STRUCTURED_OOF_PATH}"
)

print(
    f"TF-IDF            : {TFIDF_OOF_PATH}"
)


# ==============================================================================
# 2. LOAD
# ==============================================================================

modernbert_oof = pd.read_parquet(
    MODERNBERT_OOF_PATH,
    engine="pyarrow",
)

structured_oof = pd.read_parquet(
    STRUCTURED_OOF_PATH,
    engine="pyarrow",
)

tfidf_oof = pd.read_parquet(
    TFIDF_OOF_PATH,
    engine="pyarrow",
)


# ==============================================================================
# 3. POPULATION
# ==============================================================================

EXPECTED_ROWS = 35_072

for name, df in [
    ("ModernBERT", modernbert_oof),
    ("Structured + Prior", structured_oof),
    ("TF-IDF", tfidf_oof),
]:

    assert len(df) == EXPECTED_ROWS, (
        f"{name}: expected {EXPECTED_ROWS:,} rows, "
        f"found {len(df):,}"
    )

print(
    f"\n35,072-row population : PASS"
)


# ==============================================================================
# 4. REQUIRED COLUMNS
# ==============================================================================

REQUIRED_COLUMNS = {
    "response_id",
    "fold",
    "target",
}


for name, df in [
    ("ModernBERT", modernbert_oof),
    ("Structured + Prior", structured_oof),
    ("TF-IDF", tfidf_oof),
]:

    missing = (
        REQUIRED_COLUMNS
        - set(df.columns)
    )

    assert not missing, (
        f"{name} missing columns: {sorted(missing)}"
    )


print(
    "Base schema contract : PASS"
)


# ==============================================================================
# 5. PREDICTION COLUMN DISCOVERY
# ==============================================================================

def find_prediction_column(
    df,
    model_name,
):

    candidates = [
        col
        for col in df.columns
        if (
            "prediction" in str(col).lower()
            or "probability" in str(col).lower()
        )
    ]

    assert len(candidates) == 1, (
        f"{model_name}: expected exactly one "
        f"prediction column, found {candidates}"
    )

    return candidates[0]


MODERNBERT_PRED_COL = find_prediction_column(
    modernbert_oof,
    "ModernBERT",
)

STRUCTURED_PRED_COL = find_prediction_column(
    structured_oof,
    "Structured + Prior",
)

TFIDF_PRED_COL = find_prediction_column(
    tfidf_oof,
    "TF-IDF",
)


print("\n" + "=" * 100)
print("PREDICTION COLUMN LOCK")
print("=" * 100)

print(
    f"ModernBERT       : {MODERNBERT_PRED_COL}"
)

print(
    f"Structured+Prior : {STRUCTURED_PRED_COL}"
)

print(
    f"TF-IDF            : {TFIDF_PRED_COL}"
)


# ==============================================================================
# 6. NUMERICAL PREDICTION CONTRACT
# ==============================================================================

for name, df, pred_col in [
    (
        "ModernBERT",
        modernbert_oof,
        MODERNBERT_PRED_COL,
    ),
    (
        "Structured + Prior",
        structured_oof,
        STRUCTURED_PRED_COL,
    ),
    (
        "TF-IDF",
        tfidf_oof,
        TFIDF_PRED_COL,
    ),
]:

    values = pd.to_numeric(
        df[pred_col],
        errors="coerce",
    )

    assert values.notna().all(), (
        f"{name}: invalid prediction values."
    )

    assert np.isfinite(values).all(), (
        f"{name}: infinite prediction values."
    )

    assert (
        (values >= 0.0)
        & (values <= 1.0)
    ).all(), (
        f"{name}: prediction outside [0,1]."
    )


print(
    "\nPrediction numerical contract : PASS"
)


# ==============================================================================
# 7. NORMALIZE KEY TYPES
# ==============================================================================

def normalize_oof_keys(df):

    out = df.copy()

    out["response_id"] = (
        out["response_id"]
        .astype(str)
    )

    out["fold"] = (
        pd.to_numeric(
            out["fold"],
            errors="raise",
        )
        .astype(int)
    )

    out["target"] = (
        pd.to_numeric(
            out["target"],
            errors="raise",
        )
        .astype(int)
    )

    return out


modernbert_oof = normalize_oof_keys(
    modernbert_oof
)

structured_oof = normalize_oof_keys(
    structured_oof
)

tfidf_oof = normalize_oof_keys(
    tfidf_oof
)


# ==============================================================================
# 8. KEY CONTRACT
#
# IMPORTANT:
# Row order is NOT part of the OOF contract.
#
# The true alignment key is:
#
#     response_id + fold + target
#
# ==============================================================================

ALIGN_KEYS = [
    "response_id",
    "fold",
    "target",
]


for name, df in [
    ("ModernBERT", modernbert_oof),
    ("Structured + Prior", structured_oof),
    ("TF-IDF", tfidf_oof),
]:

    duplicate_mask = (
        df.duplicated(
            subset=ALIGN_KEYS,
            keep=False,
        )
    )

    duplicate_count = int(
        duplicate_mask.sum()
    )

    assert duplicate_count == 0, (
        f"{name}: duplicate OOF alignment keys "
        f"detected: {duplicate_count}"
    )


print(
    "\nUnique OOF alignment keys : PASS"
)


# ==============================================================================
# 9. KEY SET EQUALITY
# ==============================================================================

modernbert_keys = set(
    map(
        tuple,
        modernbert_oof[
            ALIGN_KEYS
        ].itertuples(
            index=False,
            name=None,
        ),
    )
)

structured_keys = set(
    map(
        tuple,
        structured_oof[
            ALIGN_KEYS
        ].itertuples(
            index=False,
            name=None,
        ),
    )
)

tfidf_keys = set(
    map(
        tuple,
        tfidf_oof[
            ALIGN_KEYS
        ].itertuples(
            index=False,
            name=None,
        ),
    )
)


assert modernbert_keys == structured_keys, (
    "ModernBERT and Structured+Prior OOF key "
    "populations do not match."
)

assert modernbert_keys == tfidf_keys, (
    "ModernBERT and TF-IDF OOF key "
    "populations do not match."
)


print(
    "Exact OOF key-population equality : PASS"
)


# ==============================================================================
# 10. REORDER INTO A SINGLE CANONICAL ORDER
#
# ModernBERT OOF is used as the canonical ordering.
# Other artifacts are keyed/reindexed to it.
# ==============================================================================

canonical_index = pd.MultiIndex.from_frame(
    modernbert_oof[
        ALIGN_KEYS
    ]
)

structured_indexed = (
    structured_oof
    .set_index(ALIGN_KEYS)
    .reindex(canonical_index)
    .reset_index()
)

tfidf_indexed = (
    tfidf_oof
    .set_index(ALIGN_KEYS)
    .reindex(canonical_index)
    .reset_index()
)


# ==============================================================================
# 11. POST-ALIGNMENT CONTRACT
# ==============================================================================

assert (
    structured_indexed[
        ALIGN_KEYS
    ].equals(
        modernbert_oof[
            ALIGN_KEYS
        ].reset_index(drop=True)
    )
), (
    "Structured+Prior failed canonical reindex."
)

assert (
    tfidf_indexed[
        ALIGN_KEYS
    ].equals(
        modernbert_oof[
            ALIGN_KEYS
        ].reset_index(drop=True)
    )
), (
    "TF-IDF failed canonical reindex."
)


print(
    "Canonical OOF alignment : PASS"
)


# ==============================================================================
# 12. TARGET CONSISTENCY AFTER ALIGNMENT
# ==============================================================================

assert (
    structured_indexed["target"].tolist()
    ==
    modernbert_oof["target"].tolist()
), (
    "Structured+Prior target mismatch after alignment."
)

assert (
    tfidf_indexed["target"].tolist()
    ==
    modernbert_oof["target"].tolist()
), (
    "TF-IDF target mismatch after alignment."
)


print(
    "Post-alignment target contract : PASS"
)


# ==============================================================================
# 13. FOLD CONSISTENCY AFTER ALIGNMENT
# ==============================================================================

assert (
    structured_indexed["fold"].tolist()
    ==
    modernbert_oof["fold"].tolist()
), (
    "Structured+Prior fold mismatch after alignment."
)

assert (
    tfidf_indexed["fold"].tolist()
    ==
    modernbert_oof["fold"].tolist()
), (
    "TF-IDF fold mismatch after alignment."
)


print(
    "Post-alignment fold contract : PASS"
)


# ==============================================================================
# 14. CANONICAL ALIGNED OOF TABLE
# ==============================================================================

aligned_oof = pd.DataFrame(
    {
        "response_id": (
            modernbert_oof[
                "response_id"
            ]
            .values
        ),

        "fold": (
            modernbert_oof[
                "fold"
            ]
            .values
        ),

        "target": (
            modernbert_oof[
                "target"
            ]
            .values
        ),

        "modernbert_prediction": (
            pd.to_numeric(
                modernbert_oof[
                    MODERNBERT_PRED_COL
                ],
                errors="raise",
            )
            .values
        ),

        "structured_prediction": (
            pd.to_numeric(
                structured_indexed[
                    STRUCTURED_PRED_COL
                ],
                errors="raise",
            )
            .values
        ),

        "tfidf_prediction": (
            pd.to_numeric(
                tfidf_indexed[
                    TFIDF_PRED_COL
                ],
                errors="raise",
            )
            .values
        ),
    }
)


# ==============================================================================
# 15. FINAL ALIGNED TABLE CONTRACT
# ==============================================================================

assert len(aligned_oof) == EXPECTED_ROWS

assert (
    aligned_oof["response_id"]
    .nunique()
    == EXPECTED_ROWS
)

assert (
    aligned_oof[
        "modernbert_prediction"
    ].notna().all()
)

assert (
    aligned_oof[
        "structured_prediction"
    ].notna().all()
)

assert (
    aligned_oof[
        "tfidf_prediction"
    ].notna().all()
)


print(
    "Aligned OOF table : PASS"
)


# ==============================================================================
# 16. GLOBAL STATE FOR CELL 2
# ==============================================================================

OOF_BLEND_SCHEMA = {
    "rows": EXPECTED_ROWS,
    "alignment_keys": ALIGN_KEYS,
    "modernbert_prediction": (
        "modernbert_prediction"
    ),
    "structured_prediction": (
        "structured_prediction"
    ),
    "tfidf_prediction": (
        "tfidf_prediction"
    ),
    "blend_weights": (
        LOCKED_BLEND_WEIGHTS.copy()
    ),
}


# ==============================================================================
# 17. FINAL STATUS
# ==============================================================================

print("\n" + "=" * 100)
print("TRACE THE ACE — CELL 1 FINAL STATUS")
print("=" * 100)

print(
    "ModernBERT OOF discovery       : PASS"
)

print(
    "Structured + Prior discovery  : PASS"
)

print(
    "TF-IDF OOF discovery           : PASS"
)

print(
    "OOF population                 : PASS"
)

print(
    "Unique alignment keys          : PASS"
)

print(
    "Key population equality        : PASS"
)

print(
    "Canonical response alignment   : PASS"
)

print(
    "Fold alignment                 : PASS"
)

print(
    "Target alignment               : PASS"
)

print(
    "Prediction numerical contract  : PASS"
)

print(
    "\nCELL 1 COMPLETE — PASS"
)
```

```python
# ==============================================================================
# TRACE THE ACE — OOF BLEND + CALIBRATION
# CELL 3 — 5-FOLD CROSS-FITTED CALIBRATION AUDIT
# ==============================================================================

import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.isotonic import IsotonicRegression
from sklearn.metrics import log_loss, roc_auc_score


print("=" * 100)
print("TRACE THE ACE — OOF BLEND + CALIBRATION")
print("CELL 3 — 5-FOLD CROSS-FITTED CALIBRATION AUDIT")
print("=" * 100)


# ==============================================================================
# 0. DEPENDENCY GATE
# ==============================================================================

assert "aligned_oof" in globals(), (
    "Cell 2 dependency failed: aligned_oof is missing."
)

assert "OOF_BLEND_METRICS" in globals(), (
    "Cell 2 dependency failed: OOF_BLEND_METRICS is missing."
)

assert "blend_prediction" in aligned_oof.columns, (
    "Cell 2 dependency failed: blend_prediction is missing."
)

print("\nCell 2 dependency : PASS")


# ==============================================================================
# 1. BASIC CONTRACT
# ==============================================================================

EXPECTED_ROWS = 35_072

assert len(aligned_oof) == EXPECTED_ROWS, (
    f"Unexpected OOF population: {len(aligned_oof)}"
)

assert (
    aligned_oof["response_id"].nunique()
    == EXPECTED_ROWS
), (
    "response_id uniqueness contract failed."
)

assert set(
    pd.to_numeric(
        aligned_oof["fold"],
        errors="raise",
    ).astype(int).unique()
) == {0, 1, 2, 3, 4}, (
    "Five-fold contract failed."
)


print(
    f"\nOOF rows : {len(aligned_oof):,}"
)

print(
    "Five-fold population : PASS"
)


# ==============================================================================
# 2. INPUT ARRAYS
# ==============================================================================

y = (
    pd.to_numeric(
        aligned_oof["target"],
        errors="raise",
    )
    .astype(int)
    .to_numpy()
)

raw_blend = (
    pd.to_numeric(
        aligned_oof["blend_prediction"],
        errors="raise",
    )
    .astype(float)
    .to_numpy()
)

folds = (
    pd.to_numeric(
        aligned_oof["fold"],
        errors="raise",
    )
    .astype(int)
    .to_numpy()
)


assert np.isfinite(raw_blend).all()

assert (
    (raw_blend >= 0.0)
    & (raw_blend <= 1.0)
).all()

print(
    "Raw blend numerical contract : PASS"
)


# ==============================================================================
# 3. RAW BLEND BASELINE
# ==============================================================================

RAW_BLEND_LL = float(
    log_loss(
        y,
        raw_blend,
        labels=[0, 1],
    )
)

RAW_BLEND_AUC = float(
    roc_auc_score(
        y,
        raw_blend,
    )
)


print("\n" + "=" * 100)
print("RAW BLEND BASELINE")
print("=" * 100)

print(
    f"Log Loss : {RAW_BLEND_LL:.12f}"
)

print(
    f"ROC-AUC  : {RAW_BLEND_AUC:.12f}"
)


# ==============================================================================
# 4. STORAGE FOR CROSS-FITTED PREDICTIONS
# ==============================================================================

crossfit_platt = np.full(
    EXPECTED_ROWS,
    np.nan,
    dtype=np.float64,
)

crossfit_isotonic = np.full(
    EXPECTED_ROWS,
    np.nan,
    dtype=np.float64,
)


# ==============================================================================
# 5. FIVE-FOLD CROSS-FITTED CALIBRATION
# ==============================================================================

print("\n" + "=" * 100)
print("5-FOLD CROSS-FITTED CALIBRATION")
print("=" * 100)


for validation_fold in [0, 1, 2, 3, 4]:

    print(
        f"\nCalibration fold {validation_fold}..."
    )

    train_mask = (
        folds != validation_fold
    )

    valid_mask = (
        folds == validation_fold
    )

    train_count = int(
        train_mask.sum()
    )

    valid_count = int(
        valid_mask.sum()
    )

    assert train_count + valid_count == EXPECTED_ROWS

    assert valid_count > 0

    print(
        f"train={train_count:,} | "
        f"valid={valid_count:,}"
    )


    # --------------------------------------------------------------------------
    # TRAIN / VALID DATA
    # --------------------------------------------------------------------------

    x_train = raw_blend[
        train_mask
    ]

    y_train = y[
        train_mask
    ]

    x_valid = raw_blend[
        valid_mask
    ]

    y_valid = y[
        valid_mask
    ]


    # ==========================================================================
    # PLATT / LOGISTIC CALIBRATION
    # ==========================================================================

    # LogisticRegression expects 2D X.
    platt_model = LogisticRegression(
        solver="lbfgs",
        max_iter=1000,
        random_state=42,
    )

    platt_model.fit(
        x_train.reshape(-1, 1),
        y_train,
    )

    platt_valid = (
        platt_model.predict_proba(
            x_valid.reshape(-1, 1)
        )[:, 1]
    )

    assert np.isfinite(
        platt_valid
    ).all()

    assert (
        (platt_valid >= 0.0)
        & (platt_valid <= 1.0)
    ).all()

    crossfit_platt[
        valid_mask
    ] = platt_valid


    # ==========================================================================
    # ISOTONIC CALIBRATION
    # ==========================================================================

    isotonic_model = IsotonicRegression(
        y_min=0.0,
        y_max=1.0,
        out_of_bounds="clip",
    )

    isotonic_model.fit(
        x_train,
        y_train,
    )

    isotonic_valid = (
        isotonic_model.predict(
            x_valid
        )
    )

    assert np.isfinite(
        isotonic_valid
    ).all()

    assert (
        (isotonic_valid >= 0.0)
        & (isotonic_valid <= 1.0)
    ).all()

    crossfit_isotonic[
        valid_mask
    ] = isotonic_valid


    print(
        f"Fold {validation_fold}: "
        f"cross-fitted calibration PASS"
    )


# ==============================================================================
# 6. CROSS-FITTED PREDICTION CONTRACT
# ==============================================================================

assert np.isfinite(
    crossfit_platt
).all(), (
    "Cross-fitted Platt contains invalid values."
)

assert np.isfinite(
    crossfit_isotonic
).all(), (
    "Cross-fitted isotonic contains invalid values."
)


assert (
    (crossfit_platt >= 0.0)
    & (crossfit_platt <= 1.0)
).all()

assert (
    (crossfit_isotonic >= 0.0)
    & (crossfit_isotonic <= 1.0)
).all()


print(
    "\nCross-fitted prediction contract : PASS"
)


# ==============================================================================
# 7. WRITE CROSS-FITTED PREDICTIONS
# ==============================================================================

aligned_oof[
    "cross_fitted_platt_prediction"
] = crossfit_platt

aligned_oof[
    "cross_fitted_isotonic_prediction"
] = crossfit_isotonic


# ==============================================================================
# 8. HONEST CROSS-FITTED METRICS
# ==============================================================================

PLATT_LL = float(
    log_loss(
        y,
        crossfit_platt,
        labels=[0, 1],
    )
)

PLATT_AUC = float(
    roc_auc_score(
        y,
        crossfit_platt,
    )
)


ISOTONIC_LL = float(
    log_loss(
        y,
        crossfit_isotonic,
        labels=[0, 1],
    )
)

ISOTONIC_AUC = float(
    roc_auc_score(
        y,
        crossfit_isotonic,
    )
)


print("\n" + "=" * 100)
print("HONEST CROSS-FITTED CALIBRATION METRICS")
print("=" * 100)

print(
    f"Raw blend       LL={RAW_BLEND_LL:.12f} "
    f"AUC={RAW_BLEND_AUC:.12f}"
)

print(
    f"Cross-fit Platt LL={PLATT_LL:.12f} "
    f"AUC={PLATT_AUC:.12f}"
)

print(
    f"Cross-fit Iso   LL={ISOTONIC_LL:.12f} "
    f"AUC={ISOTONIC_AUC:.12f}"
)


# ==============================================================================
# 9. COMPARISON TABLE
# ==============================================================================

calibration_comparison = pd.DataFrame(
    [
        {
            "method": "raw_blend",
            "log_loss": RAW_BLEND_LL,
            "roc_auc": RAW_BLEND_AUC,
        },
        {
            "method": "cross_fitted_platt",
            "log_loss": PLATT_LL,
            "roc_auc": PLATT_AUC,
        },
        {
            "method": "cross_fitted_isotonic",
            "log_loss": ISOTONIC_LL,
            "roc_auc": ISOTONIC_AUC,
        },
    ]
).sort_values(
    "log_loss",
    ascending=True,
).reset_index(drop=True)


print("\n" + "=" * 100)
print("CROSS-FITTED CALIBRATION COMPARISON")
print("=" * 100)

print(
    calibration_comparison.to_string(
        index=False
    )
)


# ==============================================================================
# 10. HONEST PRODUCTION DECISION
# ==============================================================================

best_calibrated_method = (
    calibration_comparison
    .loc[
        calibration_comparison["method"]
        != "raw_blend"
    ]
    .sort_values(
        "log_loss",
        ascending=True,
    )
    .iloc[0]["method"]
)

best_calibrated_ll = float(
    calibration_comparison
    .loc[
        calibration_comparison["method"]
        == best_calibrated_method,
        "log_loss",
    ]
    .iloc[0]
)


best_calibration_gain = (
    RAW_BLEND_LL
    - best_calibrated_ll
)


# IMPORTANT:
# Production selection uses honest cross-fitted performance.
#
# The current project lock intentionally keeps raw_blend because the
# previously audited gain from calibration was negligible.
#
# We therefore report BOTH:
#   1. statistical winner
#   2. production decision


if (
    best_calibrated_ll
    < RAW_BLEND_LL
):

    calibration_improves = True

else:

    calibration_improves = False


print("\n" + "=" * 100)
print("HONEST CALIBRATION DECISION")
print("=" * 100)

print(
    f"Raw blend LL : {RAW_BLEND_LL:.12f}"
)

print(
    f"Best CF LL   : {best_calibrated_ll:.12f}"
)

print(
    f"Best method   : {best_calibrated_method}"
)

print(
    f"LL gain       : {best_calibration_gain:.12f}"
)

print(
    f"Improves      : {calibration_improves}"
)


# ==============================================================================
# 11. CURRENT PRODUCTION POLICY
# ==============================================================================

# Current locked project decision:
# raw blend remains production method.

PRODUCTION_METHOD = "raw_blend"

assert PRODUCTION_METHOD == "raw_blend"


print("\n" + "=" * 100)
print("PRODUCTION DECISION")
print("=" * 100)

print(
    "Production method : raw_blend"
)

print(
    "Calibration deployment : DISABLED"
)

print(
    "Reason: current production lock remains raw blend."
)


# ==============================================================================
# 12. FINAL CROSS-FITTED ARTIFACT TABLE
# ==============================================================================

CALIBRATION_OOF_COLUMNS = [
    "response_id",
    "fold",
    "target",
    "blend_prediction",
    "cross_fitted_platt_prediction",
    "cross_fitted_isotonic_prediction",
]


calibration_oof = aligned_oof[
    CALIBRATION_OOF_COLUMNS
].copy()


assert len(
    calibration_oof
) == EXPECTED_ROWS

assert (
    calibration_oof[
        "response_id"
    ].nunique()
    == EXPECTED_ROWS
)

assert (
    calibration_oof[
        "cross_fitted_platt_prediction"
    ].notna().all()
)

assert (
    calibration_oof[
        "cross_fitted_isotonic_prediction"
    ].notna().all()
)


print(
    "\nCross-fitted OOF artifact : PASS"
)


# ==============================================================================
# 13. GLOBAL CALIBRATION STATE
# ==============================================================================

CALIBRATION_METRICS = {
    "rows": EXPECTED_ROWS,

    "raw_blend": {
        "log_loss": RAW_BLEND_LL,
        "roc_auc": RAW_BLEND_AUC,
    },

    "cross_fitted_platt": {
        "log_loss": PLATT_LL,
        "roc_auc": PLATT_AUC,
    },

    "cross_fitted_isotonic": {
        "log_loss": ISOTONIC_LL,
        "roc_auc": ISOTONIC_AUC,
    },

    "best_calibrated_method": (
        best_calibrated_method
    ),

    "best_calibrated_log_loss": (
        best_calibrated_ll
    ),

    "best_calibration_gain_vs_raw": (
        best_calibration_gain
    ),

    "production_method": (
        PRODUCTION_METHOD
    ),
}


# ==============================================================================
# 14. FINAL STATUS
# ==============================================================================

print("\n" + "=" * 100)
print("TRACE THE ACE — CELL 3 FINAL STATUS")
print("=" * 100)

print(
    "Cell 2 dependency             : PASS"
)

print(
    "35,072-row OOF population     : PASS"
)

print(
    "5-fold cross-fitted Platt     : PASS"
)

print(
    "5-fold cross-fitted isotonic  : PASS"
)

print(
    "No same-row calibration fit   : PASS"
)

print(
    "Cross-fitted prediction       : PASS"
)

print(
    "Calibration comparison        : PASS"
)

print(
    "Production method             : raw_blend"
)

print(
    "\nCELL 3 COMPLETE — PASS"
)
```

```python
# ==============================================================================
# TRACE THE ACE — OOF BLEND + CALIBRATION
# CELL 4 — FINAL OOF ARTIFACT + INTEGRITY GATE
# ==============================================================================

from pathlib import Path
import json
import hashlib

import numpy as np
import pandas as pd


print("=" * 100)
print("TRACE THE ACE — OOF BLEND + CALIBRATION")
print("CELL 4 — FINAL OOF ARTIFACT + INTEGRITY GATE")
print("=" * 100)


# ==============================================================================
# 0. DEPENDENCY GATE
# ==============================================================================

assert "calibration_oof" in globals(), (
    "Cell 3 dependency failed: calibration_oof is missing."
)

assert "CALIBRATION_METRICS" in globals(), (
    "Cell 3 dependency failed: CALIBRATION_METRICS is missing."
)

assert "OOF_BLEND_METRICS" in globals(), (
    "Cell 2 dependency failed: OOF_BLEND_METRICS is missing."
)

assert "LOCKED_BLEND_WEIGHTS" in globals(), (
    "Cell 0 dependency failed: locked blend weights are missing."
)

print("\nCell 3 dependency : PASS")


# ==============================================================================
# 1. FINAL ARTIFACT ROOT
# ==============================================================================

FINAL_ROOT = (
    SCRATCH_ROOT
    / "09C"
    / "blend_calibration"
)

FINAL_OUTPUT_ROOT = (
    FINAL_ROOT
    / "outputs"
)

FINAL_AUDIT_ROOT = (
    FINAL_ROOT
    / "audit"
)

FINAL_OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)

FINAL_AUDIT_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)


FINAL_OOF_PATH = (
    FINAL_OUTPUT_ROOT
    / "oof_blend_calibration.parquet"
)

FINAL_METRICS_PATH = (
    FINAL_AUDIT_ROOT
    / "cell4_final_metrics.json"
)

FINAL_MANIFEST_PATH = (
    FINAL_AUDIT_ROOT
    / "cell4_final_manifest.json"
)


print("\n" + "=" * 100)
print("FINAL ARTIFACT PATHS")
print("=" * 100)

print(
    f"OOF       : {FINAL_OOF_PATH}"
)

print(
    f"Metrics   : {FINAL_METRICS_PATH}"
)

print(
    f"Manifest  : {FINAL_MANIFEST_PATH}"
)


# ==============================================================================
# 2. FINAL PRODUCTION METHOD
# ==============================================================================

PRODUCTION_METHOD = (
    CALIBRATION_METRICS[
        "production_method"
    ]
)

assert PRODUCTION_METHOD == "raw_blend", (
    f"Unexpected production method: {PRODUCTION_METHOD}"
)

print(
    f"\nProduction method : {PRODUCTION_METHOD}"
)


# ==============================================================================
# 3. FINAL OOF SCHEMA
# ==============================================================================

FINAL_OOF_COLUMNS = [
    "response_id",
    "fold",
    "target",

    "modernbert_prediction",
    "structured_prediction",
    "tfidf_prediction",

    "blend_prediction",

    "cross_fitted_platt_prediction",
    "cross_fitted_isotonic_prediction",
]


# calibration_oof currently contains:
#
# response_id
# fold
# target
# blend_prediction
# cross_fitted_platt_prediction
# cross_fitted_isotonic_prediction
#
# Add the three component predictions from aligned_oof.

assert "aligned_oof" in globals()

final_oof = calibration_oof.copy()


for column in [
    "modernbert_prediction",
    "structured_prediction",
    "tfidf_prediction",
]:

    assert column in aligned_oof.columns

    final_oof[column] = (
        aligned_oof[column].to_numpy()
    )


final_oof = final_oof[
    FINAL_OOF_COLUMNS
].copy()


# ==============================================================================
# 4. FINAL POPULATION CONTRACT
# ==============================================================================

EXPECTED_ROWS = 35_072

assert len(final_oof) == EXPECTED_ROWS, (
    f"Final OOF row count mismatch: "
    f"{len(final_oof)} != {EXPECTED_ROWS}"
)

assert (
    final_oof[
        "response_id"
    ].nunique()
    == EXPECTED_ROWS
), (
    "Final OOF response_id uniqueness failed."
)

print(
    f"\nRows              : {len(final_oof):,}"
)

print(
    "Unique responses  : PASS"
)


# ==============================================================================
# 5. FINAL KEY CONTRACT
# ==============================================================================

assert not final_oof.duplicated(
    subset=[
        "response_id",
        "fold",
        "target",
    ]
).any(), (
    "Duplicate final OOF alignment keys detected."
)

assert set(
    final_oof["fold"]
    .astype(int)
    .unique()
) == {0, 1, 2, 3, 4}, (
    "Final OOF fold contract failed."
)

assert set(
    final_oof["target"]
    .astype(int)
    .unique()
).issubset({0, 1}), (
    "Final OOF target contract failed."
)

print(
    "Final alignment key contract : PASS"
)


# ==============================================================================
# 6. PREDICTION CONTRACT
# ==============================================================================

PREDICTION_COLUMNS = [
    "modernbert_prediction",
    "structured_prediction",
    "tfidf_prediction",
    "blend_prediction",
    "cross_fitted_platt_prediction",
    "cross_fitted_isotonic_prediction",
]


for column in PREDICTION_COLUMNS:

    values = pd.to_numeric(
        final_oof[column],
        errors="coerce",
    )

    assert values.notna().all(), (
        f"{column} contains NaN/non-numeric values."
    )

    assert np.isfinite(values).all(), (
        f"{column} contains infinite values."
    )

    assert (
        (values >= 0.0)
        & (values <= 1.0)
    ).all(), (
        f"{column} contains prediction outside [0,1]."
    )


print(
    "Prediction numerical contract : PASS"
)


# ==============================================================================
# 7. VERIFY BLEND RECOMPUTATION
# ==============================================================================

expected_blend = (
    LOCKED_BLEND_WEIGHTS["modernbert"]
    * final_oof[
        "modernbert_prediction"
    ].to_numpy()
    +
    LOCKED_BLEND_WEIGHTS["structured"]
    * final_oof[
        "structured_prediction"
    ].to_numpy()
    +
    LOCKED_BLEND_WEIGHTS["tfidf"]
    * final_oof[
        "tfidf_prediction"
    ].to_numpy()
)

observed_blend = (
    final_oof[
        "blend_prediction"
    ].to_numpy()
)

maximum_blend_difference = float(
    np.max(
        np.abs(
            expected_blend
            - observed_blend
        )
    )
)

print("\n" + "=" * 100)
print("BLEND INTEGRITY")
print("=" * 100)

print(
    f"Maximum blend difference : "
    f"{maximum_blend_difference:.12e}"
)

assert maximum_blend_difference <= 1e-12, (
    "Final blend prediction does not exactly "
    "match locked blend weights."
)

print(
    "Blend recomputation : PASS"
)


# ==============================================================================
# 8. VERIFY METRICS AGAIN
# ==============================================================================

from sklearn.metrics import (
    log_loss,
    roc_auc_score,
)


y_final = (
    final_oof[
        "target"
    ]
    .astype(int)
    .to_numpy()
)

blend_final = (
    final_oof[
        "blend_prediction"
    ]
    .astype(float)
    .to_numpy()
)


final_blend_ll = float(
    log_loss(
        y_final,
        blend_final,
        labels=[0, 1],
    )
)

final_blend_auc = float(
    roc_auc_score(
        y_final,
        blend_final,
    )
)


EXPECTED_LL = 0.543293053387
EXPECTED_AUC = 0.721930083655


ll_difference = abs(
    final_blend_ll
    - EXPECTED_LL
)

auc_difference = abs(
    final_blend_auc
    - EXPECTED_AUC
)


print("\n" + "=" * 100)
print("FINAL METRIC INTEGRITY")
print("=" * 100)

print(
    f"Observed LL : {final_blend_ll:.12f}"
)

print(
    f"Expected LL : {EXPECTED_LL:.12f}"
)

print(
    f"LL diff     : {ll_difference:.12e}"
)

print()

print(
    f"Observed AUC: {final_blend_auc:.12f}"
)

print(
    f"Expected AUC: {EXPECTED_AUC:.12f}"
)

print(
    f"AUC diff    : {auc_difference:.12e}"
)


assert ll_difference <= 1e-10, (
    "Final Log Loss does not match locked result."
)

assert auc_difference <= 1e-10, (
    "Final ROC-AUC does not match locked result."
)

print(
    "Final metric integrity : PASS"
)


# ==============================================================================
# 9. SAVE FINAL OOF
# ==============================================================================

final_oof.to_parquet(
    FINAL_OOF_PATH,
    index=False,
)

assert FINAL_OOF_PATH.exists()

print("\n" + "=" * 100)
print("FINAL OOF SERIALIZATION")
print("=" * 100)

print(
    f"Path : {FINAL_OOF_PATH}"
)

print(
    "Final OOF write : PASS"
)


# ==============================================================================
# 10. RELOAD FINAL OOF
# ==============================================================================

reloaded_oof = pd.read_parquet(
    FINAL_OOF_PATH,
    engine="pyarrow",
)


assert len(reloaded_oof) == EXPECTED_ROWS

assert (
    reloaded_oof.columns.tolist()
    == FINAL_OOF_COLUMNS
)

assert (
    reloaded_oof[
        "response_id"
    ].tolist()
    ==
    final_oof[
        "response_id"
    ].tolist()
)

assert np.allclose(
    reloaded_oof[
        "blend_prediction"
    ].to_numpy(),
    final_oof[
        "blend_prediction"
    ].to_numpy(),
    rtol=0.0,
    atol=1e-15,
)

print(
    "Final OOF reload : PASS"
)


# ==============================================================================
# 11. SHA-256
# ==============================================================================

sha256 = hashlib.sha256()

with open(
    FINAL_OOF_PATH,
    "rb",
) as file:

    for chunk in iter(
        lambda: file.read(1024 * 1024),
        b"",
    ):

        sha256.update(chunk)


FINAL_OOF_SHA256 = sha256.hexdigest()


print("\n" + "=" * 100)
print("FINAL OOF SHA-256")
print("=" * 100)

print(
    FINAL_OOF_SHA256
)


# ==============================================================================
# 12. METRICS JSON
# ==============================================================================

final_metrics = {
    "artifact": "oof_blend_calibration",

    "rows": EXPECTED_ROWS,

    "production_method": PRODUCTION_METHOD,

    "blend_weights": {
        "modernbert": float(
            LOCKED_BLEND_WEIGHTS[
                "modernbert"
            ]
        ),
        "structured": float(
            LOCKED_BLEND_WEIGHTS[
                "structured"
            ]
        ),
        "tfidf": float(
            LOCKED_BLEND_WEIGHTS[
                "tfidf"
            ]
        ),
    },

    "raw_blend": {
        "log_loss": final_blend_ll,
        "roc_auc": final_blend_auc,
    },

    "cross_fitted_calibration": {
        "platt": {
            "log_loss": float(
                CALIBRATION_METRICS[
                    "cross_fitted_platt"
                ]["log_loss"]
            ),
            "roc_auc": float(
                CALIBRATION_METRICS[
                    "cross_fitted_platt"
                ]["roc_auc"]
            ),
        },

        "isotonic": {
            "log_loss": float(
                CALIBRATION_METRICS[
                    "cross_fitted_isotonic"
                ]["log_loss"]
            ),
            "roc_auc": float(
                CALIBRATION_METRICS[
                    "cross_fitted_isotonic"
                ]["roc_auc"]
            ),
        },
    },

    "selected_calibration_method": None,

    "calibration_deployed": False,

    "oof_sha256": FINAL_OOF_SHA256,
}


with open(
    FINAL_METRICS_PATH,
    "w",
    encoding="utf-8",
) as file:

    json.dump(
        final_metrics,
        file,
        indent=2,
        sort_keys=True,
    )


assert FINAL_METRICS_PATH.exists()

print(
    "\nMetrics JSON write : PASS"
)


# ==============================================================================
# 13. MANIFEST
# ==============================================================================

manifest = {
    "cell": "cell4",
    "status": "PASS",

    "artifact_root": str(
        FINAL_ROOT
    ),

    "final_oof_path": str(
        FINAL_OOF_PATH
    ),

    "metrics_path": str(
        FINAL_METRICS_PATH
    ),

    "production_method": PRODUCTION_METHOD,

    "rows": EXPECTED_ROWS,

    "blend_weights": {
        key: float(value)
        for key, value
        in LOCKED_BLEND_WEIGHTS.items()
    },

    "raw_blend_log_loss": final_blend_ll,

    "raw_blend_roc_auc": final_blend_auc,

    "oof_sha256": FINAL_OOF_SHA256,

    "alignment_keys": [
        "response_id",
        "fold",
        "target",
    ],

    "calibration_deployed": False,

    "calibration_methods_audited": [
        "cross_fitted_platt",
        "cross_fitted_isotonic",
    ],
}


with open(
    FINAL_MANIFEST_PATH,
    "w",
    encoding="utf-8",
) as file:

    json.dump(
        manifest,
        file,
        indent=2,
        sort_keys=True,
    )


assert FINAL_MANIFEST_PATH.exists()

print(
    "Manifest write : PASS"
)


# ==============================================================================
# 14. RELOAD JSON CONTRACT
# ==============================================================================

with open(
    FINAL_METRICS_PATH,
    "r",
    encoding="utf-8",
) as file:

    reloaded_metrics = json.load(file)


with open(
    FINAL_MANIFEST_PATH,
    "r",
    encoding="utf-8",
) as file:

    reloaded_manifest = json.load(file)


assert (
    reloaded_metrics[
        "production_method"
    ]
    == "raw_blend"
)

assert (
    reloaded_manifest[
        "production_method"
    ]
    == "raw_blend"
)

assert (
    reloaded_metrics[
        "rows"
    ]
    == EXPECTED_ROWS
)

assert (
    reloaded_manifest[
        "rows"
    ]
    == EXPECTED_ROWS
)

assert np.isclose(
    reloaded_metrics[
        "raw_blend"
    ]["log_loss"],
    final_blend_ll,
    atol=1e-15,
)

assert np.isclose(
    reloaded_manifest[
        "raw_blend_log_loss"
    ],
    final_blend_ll,
    atol=1e-15,
)


print(
    "JSON reload : PASS"
)


# ==============================================================================
# 15. FINAL STATUS
# ==============================================================================

print("\n" + "=" * 100)
print("TRACE THE ACE — CELL 4 FINAL STATUS")
print("=" * 100)

print(
    "Cell 3 dependency          : PASS"
)

print(
    "Final OOF population       : PASS"
)

print(
    "Final schema               : PASS"
)

print(
    "Alignment key contract     : PASS"
)

print(
    "Prediction contract       : PASS"
)

print(
    "Blend recomputation        : PASS"
)

print(
    "Metric integrity           : PASS"
)

print(
    "Final OOF serialization    : PASS"
)

print(
    "Final OOF reload           : PASS"
)

print(
    "SHA-256 generation         : PASS"
)

print(
    "Metrics JSON               : PASS"
)

print(
    "Manifest                   : PASS"
)

print(
    "JSON reload                : PASS"
)

print(
    "\n" + "-" * 100
)

print(
    f"FINAL PRODUCTION METHOD : {PRODUCTION_METHOD}"
)

print(
    f"FINAL OOF LOG LOSS      : {final_blend_ll:.12f}"
)

print(
    f"FINAL OOF ROC-AUC       : {final_blend_auc:.12f}"
)

print(
    f"FINAL OOF ROWS          : {EXPECTED_ROWS:,}"
)

print(
    "-" * 100
)

print(
    "\nCELL 4 COMPLETE — PASS"
)
```
