```python
# =============================================================================
# TRACE THE ACE — TEST POPULATION LOCK
# CELL 1 — DATASET CANDIDATE INVENTORY
# =============================================================================

from pathlib import Path
import json
import hashlib
import os
import time

import pandas as pd


print("=" * 100)
print("TRACE THE ACE — TEST POPULATION LOCK")
print("CELL 1 — DATASET CANDIDATE INVENTORY")
print("=" * 100)


# =============================================================================
# 1. PROJECT PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"D:\Competition\Trace-the-race-local"
)

DATASET_ROOT = PROJECT_ROOT / "Dataset"

SCRATCH_ROOT = (
    PROJECT_ROOT
    / "scratch_mastery_outputs"
)

CELL_ROOT = (
    SCRATCH_ROOT
    / "09C"
    / "final_test_discovery"
    / "cell1"
)

CELL_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

INVENTORY_PATH = (
    CELL_ROOT
    / "cell1_dataset_candidate_inventory.parquet"
)

SUMMARY_PATH = (
    CELL_ROOT
    / "cell1_dataset_candidate_inventory.json"
)


assert PROJECT_ROOT.exists(), (
    f"PROJECT_ROOT not found:\n{PROJECT_ROOT}"
)

assert DATASET_ROOT.exists(), (
    f"DATASET_ROOT not found:\n{DATASET_ROOT}"
)


print("\nPROJECT ROOT")
print("-" * 80)
print(f"PROJECT_ROOT : {PROJECT_ROOT}")
print(f"Status       : PASS")

print("\nDATASET ROOT")
print("-" * 80)
print(f"DATASET_ROOT : {DATASET_ROOT}")
print(f"Status       : PASS")


# =============================================================================
# 2. CANONICAL IDENTITY TABLES
# =============================================================================
# These are ONLY used to establish identity overlap.
# They are NOT used to select a test candidate by filename.
# =============================================================================

CANONICAL_ROOT = (
    SCRATCH_ROOT
    / "01_data_foundation"
)

# Search for canonical artifacts without assuming a single historical path.
canonical_candidates = {
    "responses": list(SCRATCH_ROOT.rglob("responses.parquet")),
    "turns": list(SCRATCH_ROOT.rglob("turns.parquet")),
    "sessions": list(SCRATCH_ROOT.rglob("sessions.parquet")),
    "objectives": list(SCRATCH_ROOT.rglob("objectives.parquet")),
}


def choose_unique_canonical(name, paths):
    """
    Prefer a path that is inside a canonical/frozen/foundation directory.
    If multiple candidates remain, require explicit uniqueness.
    """
    if not paths:
        raise FileNotFoundError(
            f"Canonical {name}.parquet not found under:\n{SCRATCH_ROOT}"
        )

    preferred = [
        p for p in paths
        if any(
            token in str(p).lower()
            for token in [
                "canonical",
                "foundation",
                "data_foundation",
                "03_data_integrity",
            ]
        )
    ]

    if len(preferred) == 1:
        return preferred[0]

    if len(paths) == 1:
        return paths[0]

    raise RuntimeError(
        f"Multiple canonical candidates found for {name}:\n"
        + "\n".join(str(p) for p in paths)
    )


CANONICAL_PATHS = {
    name: choose_unique_canonical(name, paths)
    for name, paths in canonical_candidates.items()
}

print("\nCANONICAL ARTIFACTS")
print("-" * 80)

for name, path in CANONICAL_PATHS.items():
    print(f"{name:<12}: {path}")


# =============================================================================
# 3. LOAD CANONICAL IDENTITIES
# =============================================================================

responses = pd.read_parquet(
    CANONICAL_PATHS["responses"],
    columns=[
        c for c in [
            "response_id",
            "session_id",
            "objective_uid",
            "objective_raw",
            "target",
            "fold",
        ]
        if c in pd.read_parquet(
            CANONICAL_PATHS["responses"],
            engine="pyarrow",
        ).columns
    ],
)

sessions = pd.read_parquet(
    CANONICAL_PATHS["sessions"],
    columns=[
        c for c in [
            "session_id",
        ]
        if c in pd.read_parquet(
            CANONICAL_PATHS["sessions"],
            engine="pyarrow",
        ).columns
    ],
)

objectives = pd.read_parquet(
    CANONICAL_PATHS["objectives"],
)

assert "response_id" in responses.columns
assert "session_id" in responses.columns

CANONICAL_RESPONSE_IDS = set(
    responses["response_id"]
    .astype(str)
)

CANONICAL_SESSION_IDS = set(
    responses["session_id"]
    .astype(str)
)

if "objective_uid" in responses.columns:
    CANONICAL_OBJECTIVE_IDS = set(
        responses["objective_uid"]
        .dropna()
        .astype(str)
    )
else:
    CANONICAL_OBJECTIVE_IDS = set()


print("\nCANONICAL IDENTITY POPULATION")
print("-" * 80)
print(
    f"Response IDs   : {len(CANONICAL_RESPONSE_IDS):,}"
)
print(
    f"Session IDs    : {len(CANONICAL_SESSION_IDS):,}"
)
print(
    f"Objective IDs  : {len(CANONICAL_OBJECTIVE_IDS):,}"
)


assert len(CANONICAL_RESPONSE_IDS) == 35_072
assert len(CANONICAL_SESSION_IDS) == 22_821

print("Canonical identity contract : PASS")


# =============================================================================
# 4. FILE DISCOVERY
# =============================================================================

ALLOWED_EXTENSIONS = {
    ".parquet",
    ".csv",
    ".json",
    ".jsonl",
}

print("\nDATASET FILE DISCOVERY")
print("-" * 80)

all_files = []

for path in DATASET_ROOT.rglob("*"):
    if not path.is_file():
        continue

    if path.suffix.lower() not in ALLOWED_EXTENSIONS:
        continue

    all_files.append(path)

all_files = sorted(
    all_files,
    key=lambda p: str(p).lower()
)

print(
    f"Candidate files discovered : {len(all_files):,}"
)

assert len(all_files) > 0, (
    "No supported tabular/JSON files found in Dataset/"
)

print("File discovery : PASS")


# =============================================================================
# 5. HELPERS
# =============================================================================

def safe_columns_parquet(path):
    try:
        import pyarrow.parquet as pq

        return list(
            pq.ParquetFile(path)
            .schema
            .names
        )
    except Exception:
        return []


def safe_row_count_parquet(path):
    try:
        import pyarrow.parquet as pq

        return int(
            pq.ParquetFile(path)
            .metadata
            .num_rows
        )
    except Exception:
        return None


def inspect_file(path):
    suffix = path.suffix.lower()

    result = {
        "path": str(path),
        "relative_path": str(
            path.relative_to(PROJECT_ROOT)
        ),
        "filename": path.name,
        "extension": suffix,
        "size_bytes": int(path.stat().st_size),
        "rows": None,
        "columns": None,
        "column_names": None,
        "readable": False,
        "read_error": None,

        "has_response_id": False,
        "has_session_id": False,
        "has_objective_uid": False,
        "has_objective_raw": False,
        "has_target": False,
        "has_fold": False,

        "response_overlap_count": 0,
        "session_overlap_count": 0,
        "objective_overlap_count": 0,

        "response_overlap_rate": 0.0,
        "session_overlap_rate": 0.0,
        "objective_overlap_rate": 0.0,

        "target_like_columns": None,
        "identity_like_columns": None,

        "candidate_class": "UNREADABLE",
    }

    try:
        # ---------------------------------------------------------------------
        # Parquet: metadata first, avoiding full read whenever possible.
        # ---------------------------------------------------------------------
        if suffix == ".parquet":

            cols = safe_columns_parquet(path)
            rows = safe_row_count_parquet(path)

            result["columns"] = len(cols)
            result["column_names"] = cols
            result["rows"] = rows
            result["readable"] = True

            lower_to_original = {
                c.lower(): c
                for c in cols
            }

            response_col = lower_to_original.get(
                "response_id"
            )

            session_col = lower_to_original.get(
                "session_id"
            )

            objective_col = lower_to_original.get(
                "objective_uid"
            )

            target_cols = [
                c for c in cols
                if any(
                    token in c.lower()
                    for token in [
                        "target",
                        "label",
                        "answer",
                        "ground_truth",
                    ]
                )
            ]

            identity_cols = [
                c for c in cols
                if any(
                    token in c.lower()
                    for token in [
                        "response_id",
                        "session_id",
                        "objective_uid",
                        "turn_uid",
                    ]
                )
            ]

            result["has_response_id"] = (
                response_col is not None
            )

            result["has_session_id"] = (
                session_col is not None
            )

            result["has_objective_uid"] = (
                objective_col is not None
            )

            result["has_objective_raw"] = (
                "objective_raw" in lower_to_original
            )

            result["has_target"] = (
                len(target_cols) > 0
            )

            result["has_fold"] = (
                "fold" in lower_to_original
            )

            result["target_like_columns"] = (
                target_cols
            )

            result["identity_like_columns"] = (
                identity_cols
            )

            # -----------------------------------------------------------------
            # Only load identity columns for overlap calculation.
            # -----------------------------------------------------------------

            read_cols = []

            if response_col:
                read_cols.append(response_col)

            if session_col:
                read_cols.append(session_col)

            if objective_col:
                read_cols.append(objective_col)

            if read_cols:

                df_ids = pd.read_parquet(
                    path,
                    columns=read_cols,
                )

                if response_col:
                    vals = set(
                        df_ids[response_col]
                        .dropna()
                        .astype(str)
                    )

                    overlap = (
                        vals
                        & CANONICAL_RESPONSE_IDS
                    )

                    result[
                        "response_overlap_count"
                    ] = len(overlap)

                    result[
                        "response_overlap_rate"
                    ] = (
                        len(overlap) / len(vals)
                        if vals
                        else 0.0
                    )

                if session_col:
                    vals = set(
                        df_ids[session_col]
                        .dropna()
                        .astype(str)
                    )

                    overlap = (
                        vals
                        & CANONICAL_SESSION_IDS
                    )

                    result[
                        "session_overlap_count"
                    ] = len(overlap)

                    result[
                        "session_overlap_rate"
                    ] = (
                        len(overlap) / len(vals)
                        if vals
                        else 0.0
                    )

                if objective_col:
                    vals = set(
                        df_ids[objective_col]
                        .dropna()
                        .astype(str)
                    )

                    overlap = (
                        vals
                        & CANONICAL_OBJECTIVE_IDS
                    )

                    result[
                        "objective_overlap_count"
                    ] = len(overlap)

                    result[
                        "objective_overlap_rate"
                    ] = (
                        len(overlap) / len(vals)
                        if vals
                        else 0.0
                    )

        # ---------------------------------------------------------------------
        # CSV / JSON / JSONL
        # ---------------------------------------------------------------------

        elif suffix == ".csv":

            header = pd.read_csv(
                path,
                nrows=0,
            )

            cols = list(header.columns)

            result["columns"] = len(cols)
            result["column_names"] = cols
            result["readable"] = True

            lower_to_original = {
                c.lower(): c
                for c in cols
            }

            result["has_response_id"] = (
                "response_id" in lower_to_original
            )

            result["has_session_id"] = (
                "session_id" in lower_to_original
            )

            result["has_objective_uid"] = (
                "objective_uid" in lower_to_original
            )

            result["has_objective_raw"] = (
                "objective_raw" in lower_to_original
            )

            target_cols = [
                c for c in cols
                if any(
                    token in c.lower()
                    for token in [
                        "target",
                        "label",
                        "answer",
                        "ground_truth",
                    ]
                )
            ]

            result["has_target"] = (
                len(target_cols) > 0
            )

            result["has_fold"] = (
                "fold" in lower_to_original
            )

            result["target_like_columns"] = target_cols

            result["identity_like_columns"] = [
                c for c in cols
                if any(
                    token in c.lower()
                    for token in [
                        "response_id",
                        "session_id",
                        "objective_uid",
                        "turn_uid",
                    ]
                )
            ]

            # Row count only; identity overlap can be evaluated later
            # for shortlisted candidates.

            result["rows"] = int(
                sum(
                    1
                    for _ in open(
                        path,
                        "rb",
                    )
                )
                - 1
            )

        else:
            # JSON/JSONL: lightweight structural inspection.
            with open(
                path,
                "r",
                encoding="utf-8",
                errors="replace",
            ) as f:

                first_nonempty = None

                for line in f:
                    line = line.strip()

                    if line:
                        first_nonempty = line
                        break

            if first_nonempty:

                if suffix == ".jsonl":
                    obj = json.loads(
                        first_nonempty
                    )

                else:
                    with open(
                        path,
                        "r",
                        encoding="utf-8",
                        errors="replace",
                    ) as f:
                        data = json.load(f)

                    if isinstance(data, dict):
                        obj = data
                    elif isinstance(data, list) and data:
                        obj = data[0]
                    else:
                        obj = {}

                if isinstance(obj, dict):

                    cols = list(
                        obj.keys()
                    )

                    result["columns"] = len(cols)
                    result["column_names"] = cols
                    result["readable"] = True

                    lower_to_original = {
                        c.lower(): c
                        for c in cols
                    }

                    result[
                        "has_response_id"
                    ] = (
                        "response_id"
                        in lower_to_original
                    )

                    result[
                        "has_session_id"
                    ] = (
                        "session_id"
                        in lower_to_original
                    )

                    result[
                        "has_objective_uid"
                    ] = (
                        "objective_uid"
                        in lower_to_original
                    )

                    result[
                        "has_objective_raw"
                    ] = (
                        "objective_raw"
                        in lower_to_original
                    )

                    result["has_target"] = any(
                        any(
                            token in c.lower()
                            for token in [
                                "target",
                                "label",
                                "answer",
                                "ground_truth",
                            ]
                        )
                        for c in cols
                    )

                    result["has_fold"] = (
                        "fold"
                        in lower_to_original
                    )

                    result[
                        "target_like_columns"
                    ] = [
                        c for c in cols
                        if any(
                            token in c.lower()
                            for token in [
                                "target",
                                "label",
                                "answer",
                                "ground_truth",
                            ]
                        )
                    ]

                    result[
                        "identity_like_columns"
                    ] = [
                        c for c in cols
                        if any(
                            token in c.lower()
                            for token in [
                                "response_id",
                                "session_id",
                                "objective_uid",
                                "turn_uid",
                            ]
                        )
                    ]

    except Exception as exc:
        result["read_error"] = repr(exc)
        result["readable"] = False

    # -------------------------------------------------------------------------
    # Conservative classification.
    #
    # IMPORTANT:
    # This is NOT the final test lock.
    # -------------------------------------------------------------------------

    if not result["readable"]:
        result["candidate_class"] = "UNREADABLE"

    elif result["has_target"]:
        result["candidate_class"] = "LIKELY_TRAINING_OR_LABELLED"

    elif (
        result["has_response_id"]
        and result["has_session_id"]
        and result["has_objective_uid"]
    ):
        result["candidate_class"] = "IDENTITY_CANDIDATE"

    elif (
        result["has_response_id"]
        or result["has_session_id"]
        or result["has_objective_uid"]
    ):
        result["candidate_class"] = "PARTIAL_IDENTITY"

    else:
        result["candidate_class"] = "NON_IDENTITY"

    return result


# =============================================================================
# 6. RUN INVENTORY
# =============================================================================

print("\nRUNNING FILE INSPECTION")
print("-" * 80)

start_time = time.time()

records = []

for idx, path in enumerate(all_files, start=1):

    records.append(
        inspect_file(path)
    )

    if idx % 500 == 0:
        elapsed = time.time() - start_time

        print(
            f"Inspected {idx:,} / {len(all_files):,}"
            f" | elapsed={elapsed:.1f}s"
        )

inventory = pd.DataFrame(records)

print(
    f"\nInspection complete : "
    f"{len(inventory):,} files"
)

print(
    f"Elapsed              : "
    f"{time.time() - start_time:.1f}s"
)


# =============================================================================
# 7. INVENTORY INTEGRITY
# =============================================================================

assert len(inventory) == len(all_files)

assert inventory["path"].is_unique, (
    "Duplicate inventory paths detected."
)

assert inventory["readable"].sum() > 0

print("\nINVENTORY CONTRACT")
print("-" * 80)
print(
    f"Files discovered : {len(all_files):,}"
)
print(
    f"Files inspected  : {len(inventory):,}"
)
print(
    f"Readable         : "
    f"{int(inventory['readable'].sum()):,}"
)
print(
    f"Unreadable       : "
    f"{int((~inventory['readable']).sum()):,}"
)

print("Inventory contract : PASS")


# =============================================================================
# 8. CANDIDATE SUMMARY
# =============================================================================

class_counts = (
    inventory["candidate_class"]
    .value_counts()
    .rename_axis("candidate_class")
    .reset_index(name="files")
)

print("\nCANDIDATE CLASS SUMMARY")
print("-" * 80)

print(
    class_counts.to_string(
        index=False
    )
)


# =============================================================================
# 9. IDENTITY CANDIDATES
# =============================================================================

identity_candidates = inventory[
    inventory["candidate_class"]
    .isin([
        "IDENTITY_CANDIDATE",
        "PARTIAL_IDENTITY",
    ])
].copy()

identity_candidates = identity_candidates.sort_values(
    [
        "has_response_id",
        "has_session_id",
        "has_objective_uid",
        "has_target",
        "rows",
    ],
    ascending=[
        False,
        False,
        False,
        True,
        False,
    ],
)

print("\nIDENTITY CANDIDATES")
print("-" * 80)

print(
    f"Identity candidates : "
    f"{len(identity_candidates):,}"
)

display_columns = [
    "relative_path",
    "rows",
    "columns",
    "has_response_id",
    "has_session_id",
    "has_objective_uid",
    "has_target",
    "has_fold",
    "response_overlap_count",
    "session_overlap_count",
    "objective_overlap_count",
    "candidate_class",
]

available_display_columns = [
    c for c in display_columns
    if c in identity_candidates.columns
]

if len(identity_candidates):
    display(
        identity_candidates[
            available_display_columns
        ].head(100)
    )


# =============================================================================
# 10. POSSIBLE TEST CANDIDATE VIEW
# =============================================================================
# This is deliberately conservative.
#
# A file is NOT locked as test merely because it:
#   - has no target
#   - has response_id
#
# This view is only for Cell 2 forensic analysis.
# =============================================================================

possible_test = inventory[
    (
        inventory["readable"]
        &
        (~inventory["has_target"])
        &
        (
            inventory["has_response_id"]
            |
            inventory["has_session_id"]
            |
            inventory["has_objective_uid"]
        )
    )
].copy()

possible_test["identity_score"] = (
    possible_test["has_response_id"].astype(int)
    + possible_test["has_session_id"].astype(int)
    + possible_test["has_objective_uid"].astype(int)
)

possible_test = possible_test.sort_values(
    [
        "identity_score",
        "response_overlap_count",
        "session_overlap_count",
        "rows",
    ],
    ascending=[
        False,
        False,
        False,
        False,
    ],
)

print("\nPOSSIBLE TEST CANDIDATES")
print("-" * 80)

print(
    f"Possible candidates : "
    f"{len(possible_test):,}"
)

if len(possible_test):

    display(
        possible_test[
            [
                "relative_path",
                "rows",
                "columns",
                "has_response_id",
                "has_session_id",
                "has_objective_uid",
                "has_target",
                "response_overlap_count",
                "session_overlap_count",
                "objective_overlap_count",
                "identity_score",
                "candidate_class",
            ]
        ].head(100)
    )


# =============================================================================
# 11. WRITE INVENTORY ARTIFACT
# =============================================================================

inventory.to_parquet(
    INVENTORY_PATH,
    index=False,
)

assert INVENTORY_PATH.exists()

print("\nINVENTORY ARTIFACT")
print("-" * 80)
print(
    f"Inventory : {INVENTORY_PATH}"
)
print("Write : PASS")


# =============================================================================
# 12. WRITE SUMMARY
# =============================================================================

summary = {
    "cell": "12_test_population_lock_cell1",
    "status": "PASS",

    "project_root": str(PROJECT_ROOT),
    "dataset_root": str(DATASET_ROOT),

    "files_discovered": int(
        len(all_files)
    ),

    "files_inspected": int(
        len(inventory)
    ),

    "readable_files": int(
        inventory["readable"].sum()
    ),

    "unreadable_files": int(
        (~inventory["readable"]).sum()
    ),

    "identity_candidates": int(
        len(identity_candidates)
    ),

    "possible_test_candidates": int(
        len(possible_test)
    ),

    "canonical_population": {
        "responses": len(
            CANONICAL_RESPONSE_IDS
        ),
        "sessions": len(
            CANONICAL_SESSION_IDS
        ),
        "objectives": len(
            CANONICAL_OBJECTIVE_IDS
        ),
    },

    "test_population_locked": False,

    "production_inference_started": False,

    "submission_generated": False,

    "next_cell": (
        "Cell 2 — candidate identity overlap "
        "and structural forensic audit"
    ),

    "inventory_artifact": str(
        INVENTORY_PATH
    ),
}

with open(
    SUMMARY_PATH,
    "w",
    encoding="utf-8",
) as f:
    json.dump(
        summary,
        f,
        indent=2,
    )


assert SUMMARY_PATH.exists()

print(
    f"Summary    : {SUMMARY_PATH}"
)
print("Summary write : PASS")


# =============================================================================
# 13. FINAL STATUS
# =============================================================================

print("\n" + "=" * 100)
print("TRACE THE ACE — CELL 1 FINAL STATUS")
print("=" * 100)

print(
    f"Dataset discovery          : PASS"
)

print(
    f"File inspection            : PASS"
)

print(
    f"Identity candidate audit   : PASS"
)

print(
    f"Possible test candidates  : "
    f"{len(possible_test):,}"
)

print(
    "Test population lock       : NOT YET"
)

print(
    "Production inference       : NOT STARTED"
)

print(
    "Submission generation      : NOT STARTED"
)

print("-" * 100)

print(
    "CELL 1 COMPLETE — PASS"
)

print(
    "IMPORTANT: No test population was locked in Cell 1."
)
print(
    "Cell 2 must perform candidate identity/structural forensic audit."
)

print("=" * 100)
```

```python
# =============================================================================
# TRACE THE ACE — TEST POPULATION LOCK
# CELL 2 — EXACT IDENTITY OVERLAP + CANDIDATE FORENSIC AUDIT
# =============================================================================

from pathlib import Path
import json
import numpy as np
import pandas as pd


print("=" * 100)
print("TRACE THE ACE — TEST POPULATION LOCK")
print("CELL 2 — EXACT IDENTITY OVERLAP + CANDIDATE FORENSIC AUDIT")
print("=" * 100)


# =============================================================================
# 0. DEPENDENCY GATE
# =============================================================================

assert "inventory" in globals(), (
    "Cell 1 dependency failed: inventory is missing."
)

assert "responses" in globals(), (
    "Cell 1 dependency failed: responses is missing."
)

assert "CANONICAL_RESPONSE_IDS" in globals(), (
    "Cell 1 dependency failed: canonical response IDs are missing."
)

assert "CANONICAL_SESSION_IDS" in globals(), (
    "Cell 1 dependency failed: canonical session IDs are missing."
)

assert "CANONICAL_OBJECTIVE_IDS" in globals(), (
    "Cell 1 dependency failed: canonical objective IDs are missing."
)


print("\nCell 1 dependency : PASS")


# =============================================================================
# 1. OUTPUT PATHS
# =============================================================================

CELL_ROOT = (
    SCRATCH_ROOT
    / "09C"
    / "final_test_discovery"
    / "cell2"
)

CELL_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)

FORENSIC_PATH = (
    CELL_ROOT
    / "cell2_identity_forensic.parquet"
)

SUMMARY_PATH = (
    CELL_ROOT
    / "cell2_identity_forensic.json"
)

SHORTLIST_PATH = (
    CELL_ROOT
    / "cell2_candidate_shortlist.parquet"
)


# =============================================================================
# 2. CANONICAL POPULATION
# =============================================================================

CANONICAL_RESPONSE_COUNT = len(
    CANONICAL_RESPONSE_IDS
)

CANONICAL_SESSION_COUNT = len(
    CANONICAL_SESSION_IDS
)

CANONICAL_OBJECTIVE_COUNT = len(
    CANONICAL_OBJECTIVE_IDS
)


assert CANONICAL_RESPONSE_COUNT == 35_072
assert CANONICAL_SESSION_COUNT == 22_821
assert CANONICAL_OBJECTIVE_COUNT == 398


print("\n" + "=" * 100)
print("CANONICAL POPULATION")
print("=" * 100)

print(
    f"Responses  : {CANONICAL_RESPONSE_COUNT:,}"
)

print(
    f"Sessions   : {CANONICAL_SESSION_COUNT:,}"
)

print(
    f"Objectives : {CANONICAL_OBJECTIVE_COUNT:,}"
)

print(
    "Canonical population : PASS"
)


# =============================================================================
# 3. IDENTITY COLUMN EXTRACTION
# =============================================================================

def get_identity_columns(path):
    """
    Return the actual source columns that correspond to the canonical
    identity fields.

    Case-insensitive matching is allowed.
    No semantic guessing beyond exact field names.
    """

    columns = []

    try:

        suffix = path.suffix.lower()

        if suffix == ".parquet":

            import pyarrow.parquet as pq

            columns = (
                pq.ParquetFile(path)
                .schema
                .names
            )

        elif suffix == ".csv":

            columns = list(
                pd.read_csv(
                    path,
                    nrows=0,
                ).columns
            )

        else:

            return {}

    except Exception:

        return {}

    lower_map = {
        str(c).lower(): c
        for c in columns
    }

    return {
        "response_id": lower_map.get(
            "response_id"
        ),
        "session_id": lower_map.get(
            "session_id"
        ),
        "objective_uid": lower_map.get(
            "objective_uid"
        ),
    }


# =============================================================================
# 4. EXACT OVERLAP AUDIT
# =============================================================================

def audit_identity_overlap(path):
    """
    Read ONLY exact identity columns.

    This deliberately avoids loading arbitrary feature columns.
    """

    result = {
        "response_overlap_count": 0,
        "response_overlap_rate": 0.0,

        "session_overlap_count": 0,
        "session_overlap_rate": 0.0,

        "objective_overlap_count": 0,
        "objective_overlap_rate": 0.0,

        "response_unique_count": 0,
        "session_unique_count": 0,
        "objective_unique_count": 0,

        "identity_readable": False,
        "identity_error": None,
    }

    try:

        identity_cols = get_identity_columns(
            path
        )

        read_columns = [
            column
            for column in identity_cols.values()
            if column is not None
        ]

        if not read_columns:

            return result

        suffix = path.suffix.lower()

        if suffix == ".parquet":

            df = pd.read_parquet(
                path,
                columns=read_columns,
                engine="pyarrow",
            )

        elif suffix == ".csv":

            df = pd.read_csv(
                path,
                usecols=read_columns,
            )

        else:

            return result


        result["identity_readable"] = True


        # ---------------------------------------------------------------------
        # response_id
        # ---------------------------------------------------------------------

        response_col = identity_cols[
            "response_id"
        ]

        if response_col is not None:

            values = set(
                df[
                    response_col
                ]
                .dropna()
                .astype(str)
            )

            result[
                "response_unique_count"
            ] = len(values)

            overlap = (
                values
                & CANONICAL_RESPONSE_IDS
            )

            result[
                "response_overlap_count"
            ] = len(overlap)

            result[
                "response_overlap_rate"
            ] = (
                len(overlap) / len(values)
                if values
                else 0.0
            )


        # ---------------------------------------------------------------------
        # session_id
        # ---------------------------------------------------------------------

        session_col = identity_cols[
            "session_id"
        ]

        if session_col is not None:

            values = set(
                df[
                    session_col
                ]
                .dropna()
                .astype(str)
            )

            result[
                "session_unique_count"
            ] = len(values)

            overlap = (
                values
                & CANONICAL_SESSION_IDS
            )

            result[
                "session_overlap_count"
            ] = len(overlap)

            result[
                "session_overlap_rate"
            ] = (
                len(overlap) / len(values)
                if values
                else 0.0
            )


        # ---------------------------------------------------------------------
        # objective_uid
        # ---------------------------------------------------------------------

        objective_col = identity_cols[
            "objective_uid"
        ]

        if objective_col is not None:

            values = set(
                df[
                    objective_col
                ]
                .dropna()
                .astype(str)
            )

            result[
                "objective_unique_count"
            ] = len(values)

            overlap = (
                values
                & CANONICAL_OBJECTIVE_IDS
            )

            result[
                "objective_overlap_count"
            ] = len(overlap)

            result[
                "objective_overlap_rate"
            ] = (
                len(overlap) / len(values)
                if values
                else 0.0
            )

    except Exception as exc:

        result[
            "identity_error"
        ] = repr(exc)

    return result


# =============================================================================
# 5. SELECT INVENTORY CANDIDATES
# =============================================================================
# Cell 1 found every Dataset file to be PARTIAL_IDENTITY.
#
# We now audit only files which contain at least one exact identity field.
# =============================================================================

identity_inventory = inventory[
    (
        inventory["has_response_id"]
        |
        inventory["has_session_id"]
        |
        inventory["has_objective_uid"]
    )
].copy()

identity_inventory = (
    identity_inventory
    .reset_index(drop=True)
)


print("\n" + "=" * 100)
print("IDENTITY AUDIT INPUT")
print("=" * 100)

print(
    f"Identity-bearing files : "
    f"{len(identity_inventory):,}"
)

assert len(identity_inventory) > 0

print(
    "Identity audit input : PASS"
)


# =============================================================================
# 6. RUN EXACT OVERLAP AUDIT
# =============================================================================

print("\n" + "=" * 100)
print("RUNNING EXACT IDENTITY OVERLAP AUDIT")
print("=" * 100)


audit_records = []

for idx, row in identity_inventory.iterrows():

    path = Path(
        row["path"]
    )

    overlap_result = (
        audit_identity_overlap(
            path
        )
    )

    record = row.to_dict()

    record.update(
        overlap_result
    )

    audit_records.append(
        record
    )

    if (idx + 1) % 500 == 0:

        print(
            f"Audited "
            f"{idx + 1:,} / "
            f"{len(identity_inventory):,}"
        )


identity_forensic = pd.DataFrame(
    audit_records
)


assert len(
    identity_forensic
) == len(
    identity_inventory
)


print(
    "\nExact identity audit complete : PASS"
)


# =============================================================================
# 7. OVERLAP CLASSIFICATION
# =============================================================================

def classify_overlap(row):

    if not row["identity_readable"]:
        return "IDENTITY_UNREADABLE"

    response_overlap = int(
        row["response_overlap_count"]
    )

    session_overlap = int(
        row["session_overlap_count"]
    )

    objective_overlap = int(
        row["objective_overlap_count"]
    )

    has_response = bool(
        row["has_response_id"]
    )

    has_session = bool(
        row["has_session_id"]
    )

    has_objective = bool(
        row["has_objective_uid"]
    )

    has_target = bool(
        row["has_target"]
    )


    # -------------------------------------------------------------------------
    # Any target-bearing identity file is not a clean unseen-test candidate.
    # -------------------------------------------------------------------------

    if has_target:

        return "LABELLED_OR_TRAINING"


    # -------------------------------------------------------------------------
    # Exact full response overlap.
    # -------------------------------------------------------------------------

    if (
        has_response
        and
        response_overlap
        == row["response_unique_count"]
        and
        response_overlap > 0
    ):

        return "FULL_RESPONSE_OVERLAP"


    # -------------------------------------------------------------------------
    # Exact full session overlap.
    # -------------------------------------------------------------------------

    if (
        has_session
        and
        session_overlap
        == row["session_unique_count"]
        and
        session_overlap > 0
    ):

        return "FULL_SESSION_OVERLAP"


    # -------------------------------------------------------------------------
    # Partial canonical overlap.
    # -------------------------------------------------------------------------

    if (
        response_overlap > 0
        or session_overlap > 0
        or objective_overlap > 0
    ):

        return "PARTIAL_CANONICAL_OVERLAP"


    # -------------------------------------------------------------------------
    # Zero canonical identity overlap.
    # -------------------------------------------------------------------------

    if (
        response_overlap == 0
        and
        session_overlap == 0
        and
        objective_overlap == 0
    ):

        return "ZERO_CANONICAL_OVERLAP"


    return "AMBIGUOUS"


identity_forensic[
    "overlap_class"
] = identity_forensic.apply(
    classify_overlap,
    axis=1,
)


# =============================================================================
# 8. FORENSIC SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("IDENTITY OVERLAP SUMMARY")
print("=" * 100)

overlap_summary = (
    identity_forensic[
        "overlap_class"
    ]
    .value_counts()
    .rename_axis(
        "overlap_class"
    )
    .reset_index(
        name="files"
    )
)

print(
    overlap_summary.to_string(
        index=False
    )
)


# =============================================================================
# 9. ZERO-OVERLAP CANDIDATES
# =============================================================================

zero_overlap = identity_forensic[
    identity_forensic[
        "overlap_class"
    ]
    ==
    "ZERO_CANONICAL_OVERLAP"
].copy()


zero_overlap = zero_overlap.sort_values(
    [
        "has_response_id",
        "has_session_id",
        "has_objective_uid",
        "rows",
    ],
    ascending=[
        False,
        False,
        False,
        False,
    ],
)


print("\n" + "=" * 100)
print("ZERO-CANONICAL-OVERLAP CANDIDATES")
print("=" * 100)

print(
    f"Zero-overlap files : "
    f"{len(zero_overlap):,}"
)


if len(zero_overlap):

    display(
        zero_overlap[
            [
                "relative_path",
                "rows",
                "columns",
                "has_response_id",
                "has_session_id",
                "has_objective_uid",
                "has_target",
                "response_unique_count",
                "session_unique_count",
                "objective_unique_count",
                "overlap_class",
            ]
        ].head(200)
    )


# =============================================================================
# 10. HIGH-INTEREST UNLABELLED CANDIDATES
# =============================================================================
# This is a shortlist only.
#
# It does NOT mean these are test rows.
# =============================================================================

high_interest = identity_forensic[
    (
        ~identity_forensic["has_target"]
    )
    &
    (
        identity_forensic[
            "identity_readable"
        ]
    )
    &
    (
        identity_forensic[
            "overlap_class"
        ].isin([
            "ZERO_CANONICAL_OVERLAP",
            "PARTIAL_CANONICAL_OVERLAP",
        ])
    )
].copy()


high_interest[
    "candidate_score"
] = 0


# Stronger identity schema
high_interest[
    "candidate_score"
] += (
    high_interest[
        "has_response_id"
    ].astype(int)
    * 5
)

high_interest[
    "candidate_score"
] += (
    high_interest[
        "has_session_id"
    ].astype(int)
    * 3
)

high_interest[
    "candidate_score"
] += (
    high_interest[
        "has_objective_uid"
    ].astype(int)
    * 2
)


# Prefer zero response overlap over partial response overlap
high_interest[
    "candidate_score"
] += (
    (
        high_interest[
            "response_overlap_count"
        ]
        == 0
    )
    .astype(int)
    * 4
)


# Penalize training-like population overlap
high_interest[
    "candidate_score"
] -= (
    high_interest[
        "response_overlap_count"
    ]
    > 0
).astype(int) * 3


high_interest = high_interest.sort_values(
    [
        "candidate_score",
        "rows",
        "has_response_id",
        "has_session_id",
    ],
    ascending=[
        False,
        False,
        False,
        False,
    ],
)


print("\n" + "=" * 100)
print("HIGH-INTEREST UNLABELLED CANDIDATES")
print("=" * 100)

print(
    f"Shortlist size : "
    f"{len(high_interest):,}"
)

if len(high_interest):

    display(
        high_interest[
            [
                "relative_path",
                "rows",
                "columns",
                "has_response_id",
                "has_session_id",
                "has_objective_uid",
                "response_overlap_count",
                "session_overlap_count",
                "objective_overlap_count",
                "candidate_score",
                "overlap_class",
            ]
        ].head(100)
    )


# =============================================================================
# 11. IMPORTANT SAFETY ASSERTIONS
# =============================================================================

# No candidate may be declared "LOCKED_TEST" here.
assert (
    "LOCKED_TEST"
    not in identity_forensic.columns
)


# This cell must not generate predictions.
assert (
    not PRODUCTION_INFERENCE_STARTED
)


# This cell must not modify canonical training artifacts.
assert (
    CANONICAL_RESPONSE_COUNT
    == 35_072
)


print("\n" + "=" * 100)
print("SAFETY CONTRACT")
print("=" * 100)

print(
    "No test population locked : PASS"
)

print(
    "No production inference   : PASS"
)

print(
    "Canonical population unchanged : PASS"
)


# =============================================================================
# 12. WRITE FORENSIC ARTIFACT
# =============================================================================

identity_forensic.to_parquet(
    FORENSIC_PATH,
    index=False,
)

assert FORENSIC_PATH.exists()

print("\nFORENSIC ARTIFACT")
print("-" * 100)
print(
    f"Path : {FORENSIC_PATH}"
)

print(
    "Forensic parquet write : PASS"
)


# =============================================================================
# 13. WRITE SHORTLIST
# =============================================================================

shortlist_columns = [
    "path",
    "relative_path",
    "filename",
    "extension",
    "rows",
    "columns",

    "has_response_id",
    "has_session_id",
    "has_objective_uid",
    "has_target",

    "response_unique_count",
    "session_unique_count",
    "objective_unique_count",

    "response_overlap_count",
    "session_overlap_count",
    "objective_overlap_count",

    "response_overlap_rate",
    "session_overlap_rate",
    "objective_overlap_rate",

    "candidate_score",
    "overlap_class",
]


shortlist_columns = [
    c for c in shortlist_columns
    if c in high_interest.columns
]


high_interest[
    shortlist_columns
].to_parquet(
    SHORTLIST_PATH,
    index=False,
)

assert SHORTLIST_PATH.exists()

print(
    f"Shortlist : {SHORTLIST_PATH}"
)

print(
    "Shortlist write : PASS"
)


# =============================================================================
# 14. WRITE SUMMARY JSON
# =============================================================================

summary = {
    "cell": "12_test_population_lock_cell2",

    "status": "PASS",

    "identity_files_audited": int(
        len(identity_forensic)
    ),

    "overlap_classes": {
        str(row["overlap_class"]): int(
            row["files"]
        )
        for _, row
        in overlap_summary.iterrows()
    },

    "zero_canonical_overlap_files": int(
        len(zero_overlap)
    ),

    "high_interest_unlabelled_candidates": int(
        len(high_interest)
    ),

    "canonical_population": {
        "responses": int(
            CANONICAL_RESPONSE_COUNT
        ),
        "sessions": int(
            CANONICAL_SESSION_COUNT
        ),
        "objectives": int(
            CANONICAL_OBJECTIVE_COUNT
        ),
    },

    "test_population_locked": False,

    "production_inference_started": False,

    "submission_generated": False,

    "forensic_artifact": str(
        FORENSIC_PATH
    ),

    "shortlist_artifact": str(
        SHORTLIST_PATH
    ),

    "next_cell": (
        "Cell 3 — candidate structural "
        "forensic comparison and test-population evidence audit"
    ),
}


with open(
    SUMMARY_PATH,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        summary,
        f,
        indent=2,
    )


assert SUMMARY_PATH.exists()

print(
    f"Summary : {SUMMARY_PATH}"
)

print(
    "Summary write : PASS"
)


# =============================================================================
# 15. FINAL STATUS
# =============================================================================

print("\n" + "=" * 100)
print("TRACE THE ACE — CELL 2 FINAL STATUS")
print("=" * 100)

print(
    "Cell 1 dependency              : PASS"
)

print(
    "Exact identity audit           : PASS"
)

print(
    "Response overlap audit         : PASS"
)

print(
    "Session overlap audit          : PASS"
)

print(
    "Objective overlap audit        : PASS"
)

print(
    "Candidate forensic shortlist   : PASS"
)

print(
    "Test population lock           : NOT YET"
)

print(
    "Production inference           : NOT STARTED"
)

print(
    "Submission generation          : NOT STARTED"
)

print("-" * 100)

print(
    "CELL 2 COMPLETE — PASS"
)

print(
    "IMPORTANT: Zero overlap is a candidate signal, "
    "NOT proof of test population."
)

print("=" * 100)
```

```python
# =============================================================================
# TRACE THE ACE — TEST POPULATION LOCK
# CELL 3 — FULL-RESPONSE OVERLAP FORENSIC +
#          DATASET POPULATION STRUCTURE AUDIT
# =============================================================================

from pathlib import Path
import json

import numpy as np
import pandas as pd


print("=" * 100)
print("TRACE THE ACE — TEST POPULATION LOCK")
print("CELL 3 — FULL-RESPONSE OVERLAP FORENSIC + POPULATION STRUCTURE AUDIT")
print("=" * 100)


# =============================================================================
# 0. DEPENDENCY GATE
# =============================================================================

assert "identity_forensic" in globals(), (
    "Cell 2 dependency failed: identity_forensic is missing."
)

assert "responses" in globals(), (
    "Cell 2 dependency failed: responses is missing."
)

assert "CANONICAL_RESPONSE_IDS" in globals(), (
    "Cell 2 dependency failed: canonical response IDs are missing."
)

assert "CANONICAL_SESSION_IDS" in globals(), (
    "Cell 2 dependency failed: canonical session IDs are missing."
)

assert "CANONICAL_OBJECTIVE_IDS" in globals(), (
    "Cell 2 dependency failed: canonical objective IDs are missing."
)


print("\nCell 2 dependency : PASS")


# =============================================================================
# 1. OUTPUT PATHS
# =============================================================================

CELL_ROOT = (
    SCRATCH_ROOT
    / "09C"
    / "final_test_discovery"
    / "cell3"
)

CELL_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)

FORENSIC_PATH = (
    CELL_ROOT
    / "cell3_full_response_overlap_forensic.parquet"
)

SUMMARY_PATH = (
    CELL_ROOT
    / "cell3_population_structure.json"
)


# =============================================================================
# 2. FULL RESPONSE OVERLAP FILES
# =============================================================================

full_response_overlap = identity_forensic[
    identity_forensic[
        "overlap_class"
    ]
    ==
    "FULL_RESPONSE_OVERLAP"
].copy()


print("\n" + "=" * 100)
print("FULL-RESPONSE-OVERLAP FILES")
print("=" * 100)

print(
    f"Files : {len(full_response_overlap):,}"
)


assert len(full_response_overlap) == 2, (
    "Expected exactly 2 FULL_RESPONSE_OVERLAP files "
    "from Cell 2."
)


display(
    full_response_overlap[
        [
            "relative_path",
            "filename",
            "extension",
            "rows",
            "columns",
            "has_response_id",
            "has_session_id",
            "has_objective_uid",
            "has_target",
            "has_fold",
            "response_unique_count",
            "response_overlap_count",
            "session_unique_count",
            "session_overlap_count",
        ]
    ]
)


# =============================================================================
# 3. LOAD EXACT IDENTITY DATA FOR THE TWO FILES
# =============================================================================

def load_identity_frame(path):

    suffix = path.suffix.lower()

    if suffix == ".parquet":

        import pyarrow.parquet as pq

        cols = pq.ParquetFile(
            path
        ).schema.names

        lower_map = {
            str(c).lower(): c
            for c in cols
        }

        selected = [
            lower_map[c]
            for c in [
                "response_id",
                "session_id",
                "objective_uid",
                "target",
                "fold",
            ]
            if c in lower_map
        ]

        return pd.read_parquet(
            path,
            columns=selected,
            engine="pyarrow",
        )

    if suffix == ".csv":

        header = pd.read_csv(
            path,
            nrows=0,
        )

        lower_map = {
            str(c).lower(): c
            for c in header.columns
        }

        selected = [
            lower_map[c]
            for c in [
                "response_id",
                "session_id",
                "objective_uid",
                "target",
                "fold",
            ]
            if c in lower_map
        ]

        return pd.read_csv(
            path,
            usecols=selected,
        )

    raise ValueError(
        f"Unsupported file type: {suffix}"
    )


full_overlap_records = []


for _, row in full_response_overlap.iterrows():

    path = Path(
        row["path"]
    )

    df = load_identity_frame(
        path
    )

    print("\n" + "-" * 100)
    print(
        f"FILE: {path}"
    )

    print(
        f"Rows loaded : {len(df):,}"
    )

    # -------------------------------------------------------------------------
    # Exact response comparison
    # -------------------------------------------------------------------------

    assert "response_id" in df.columns

    file_response_ids = set(
        df[
            "response_id"
        ]
        .dropna()
        .astype(str)
    )

    response_overlap = (
        file_response_ids
        & CANONICAL_RESPONSE_IDS
    )

    print(
        f"Unique response IDs : "
        f"{len(file_response_ids):,}"
    )

    print(
        f"Canonical overlap   : "
        f"{len(response_overlap):,}"
    )

    # -------------------------------------------------------------------------
    # Exact equality test
    # -------------------------------------------------------------------------

    exact_response_population = (
        file_response_ids
        ==
        CANONICAL_RESPONSE_IDS
    )

    print(
        f"Exact canonical response population : "
        f"{exact_response_population}"
    )

    # -------------------------------------------------------------------------
    # Session population
    # -------------------------------------------------------------------------

    if "session_id" in df.columns:

        file_session_ids = set(
            df[
                "session_id"
            ]
            .dropna()
            .astype(str)
        )

        session_overlap = (
            file_session_ids
            & CANONICAL_SESSION_IDS
        )

        exact_session_population = (
            file_session_ids
            ==
            CANONICAL_SESSION_IDS
        )

    else:

        file_session_ids = set()
        session_overlap = set()
        exact_session_population = False


    print(
        f"Unique session IDs   : "
        f"{len(file_session_ids):,}"
    )

    print(
        f"Canonical session overlap : "
        f"{len(session_overlap):,}"
    )

    print(
        f"Exact canonical session population : "
        f"{exact_session_population}"
    )


    # -------------------------------------------------------------------------
    # Target presence
    # -------------------------------------------------------------------------

    has_target = (
        "target" in df.columns
    )

    if has_target:

        target_values = (
            df[
                "target"
            ]
            .dropna()
            .unique()
        )

    else:

        target_values = []


    # -------------------------------------------------------------------------
    # Fold presence
    # -------------------------------------------------------------------------

    has_fold = (
        "fold" in df.columns
    )

    if has_fold:

        fold_values = sorted(
            pd.to_numeric(
                df["fold"],
                errors="coerce",
            )
            .dropna()
            .astype(int)
            .unique()
            .tolist()
        )

    else:

        fold_values = []


    full_overlap_records.append(
        {
            "path": str(path),

            "rows": int(len(df)),

            "unique_response_ids": int(
                len(file_response_ids)
            ),

            "canonical_response_overlap": int(
                len(response_overlap)
            ),

            "exact_response_population": bool(
                exact_response_population
            ),

            "unique_session_ids": int(
                len(file_session_ids)
            ),

            "canonical_session_overlap": int(
                len(session_overlap)
            ),

            "exact_session_population": bool(
                exact_session_population
            ),

            "has_target": bool(
                has_target
            ),

            "target_unique_values": [
                str(x)
                for x in target_values[:50]
            ],

            "has_fold": bool(
                has_fold
            ),

            "fold_values": fold_values,
        }
    )


# =============================================================================
# 4. FORENSIC TABLE
# =============================================================================

full_overlap_forensic = pd.DataFrame(
    full_overlap_records
)


assert len(
    full_overlap_forensic
) == 2


print("\n" + "=" * 100)
print("FULL-RESPONSE FORENSIC RESULT")
print("=" * 100)

display(
    full_overlap_forensic
)


# =============================================================================
# 5. DATASET POPULATION STRUCTURE
# =============================================================================

dataset_row_counts = (
    identity_forensic[
        identity_forensic[
            "readable"
        ]
    ]
    .groupby(
        "rows",
        dropna=False,
    )
    .size()
    .reset_index(
        name="files"
    )
    .sort_values(
        "files",
        ascending=False,
    )
)


print("\n" + "=" * 100)
print("DATASET POPULATION STRUCTURE")
print("=" * 100)

display(
    dataset_row_counts.head(50)
)


# =============================================================================
# 6. CANONICAL-SIZE FILES
# =============================================================================

canonical_size_files = identity_forensic[
    identity_forensic[
        "rows"
    ]
    .isin([
        35_072,
        22_821,
    ])
].copy()


print("\n" + "=" * 100)
print("CANONICAL-SIZE FILES")
print("=" * 100)

print(
    f"Files with canonical-sized populations : "
    f"{len(canonical_size_files):,}"
)


if len(canonical_size_files):

    display(
        canonical_size_files[
            [
                "relative_path",
                "rows",
                "columns",
                "has_response_id",
                "has_session_id",
                "has_objective_uid",
                "has_target",
                "overlap_class",
            ]
        ]
        .sort_values(
            [
                "rows",
                "relative_path",
            ]
        )
        .head(200)
    )


# =============================================================================
# 7. LOOK FOR ANY DATASET FILE WITH A POPULATION SIZE
#    THAT DOES NOT MATCH TRAINING STRUCTURE
# =============================================================================

# We intentionally do NOT infer test from row count.
#
# This section only reports unusual population sizes.

known_training_population_sizes = {
    35_072,
    22_821,
    398,
}


unusual_population_files = identity_forensic[
    ~identity_forensic[
        "rows"
    ].isin(
        known_training_population_sizes
    )
].copy()


print("\n" + "=" * 100)
print("UNUSUAL POPULATION-SIZE FILES")
print("=" * 100)

print(
    f"Files with non-canonical row counts : "
    f"{len(unusual_population_files):,}"
)


if len(unusual_population_files):

    display(
        unusual_population_files[
            [
                "relative_path",
                "rows",
                "columns",
                "has_response_id",
                "has_session_id",
                "has_objective_uid",
                "has_target",
                "response_overlap_count",
                "session_overlap_count",
                "objective_overlap_count",
                "overlap_class",
            ]
        ]
        .sort_values(
            [
                "rows",
                "relative_path",
            ],
            ascending=[
                False,
                True,
            ],
        )
        .head(200)
    )


# =============================================================================
# 8. HARD NEGATIVE TEST ASSERTION
# =============================================================================
# If every identity-bearing Dataset file is tied to canonical training
# population, then Dataset/ is NOT a discoverable unseen-test source.
#
# This is a conclusion about Dataset/ only.
# It does NOT claim that the competition has no test data elsewhere.

zero_overlap_count = int(
    (
        identity_forensic[
            "overlap_class"
        ]
        ==
        "ZERO_CANONICAL_OVERLAP"
    ).sum()
)

partial_overlap_count = int(
    (
        identity_forensic[
            "overlap_class"
        ]
        ==
        "PARTIAL_CANONICAL_OVERLAP"
    ).sum()
)

full_session_count = int(
    (
        identity_forensic[
            "overlap_class"
        ]
        ==
        "FULL_SESSION_OVERLAP"
    ).sum()
)

full_response_count = int(
    (
        identity_forensic[
            "overlap_class"
        ]
        ==
        "FULL_RESPONSE_OVERLAP"
    ).sum()
)


dataset_contains_zero_overlap_candidate = (
    zero_overlap_count > 0
)


dataset_contains_unseen_identity_candidate = (
    zero_overlap_count > 0
    or
    partial_overlap_count > 0
)


print("\n" + "=" * 100)
print("DATASET TEST-POPULATION EVIDENCE")
print("=" * 100)

print(
    f"Zero-overlap files       : "
    f"{zero_overlap_count:,}"
)

print(
    f"Partial-overlap files    : "
    f"{partial_overlap_count:,}"
)

print(
    f"Full-session-overlap     : "
    f"{full_session_count:,}"
)

print(
    f"Full-response-overlap    : "
    f"{full_response_count:,}"
)


if not dataset_contains_unseen_identity_candidate:

    print(
        "\nDataset unseen-identity candidate : NONE"
    )

    print(
        "Dataset test population evidence : NOT FOUND"
    )

else:

    print(
        "\nDataset unseen-identity candidate : PRESENT"
    )

    print(
        "Further forensic analysis required."
    )


# =============================================================================
# 9. NO LOCK / NO INFERENCE GUARD
# =============================================================================

TEST_POPULATION_LOCKED = False

assert (
    TEST_POPULATION_LOCKED is False
)

assert (
    not PRODUCTION_INFERENCE_STARTED
)


print("\n" + "=" * 100)
print("SAFETY GATE")
print("=" * 100)

print(
    "Test population locked : NO"
)

print(
    "Production inference   : NOT STARTED"
)

print(
    "Submission             : NOT GENERATED"
)


# =============================================================================
# 10. WRITE FORENSIC ARTIFACT
# =============================================================================

full_overlap_forensic.to_parquet(
    FORENSIC_PATH,
    index=False,
)

assert FORENSIC_PATH.exists()

print("\nForensic parquet : PASS")
print(
    f"Path : {FORENSIC_PATH}"
)


# =============================================================================
# 11. SUMMARY
# =============================================================================

dataset_test_population_status = (
    "NOT_FOUND_IN_DATASET"
    if not dataset_contains_unseen_identity_candidate
    else
    "CANDIDATES_REQUIRE_FURTHER_AUDIT"
)


summary = {
    "cell": "12_test_population_lock_cell3",

    "status": "PASS",

    "dataset_test_population_status": (
        dataset_test_population_status
    ),

    "identity_files_audited": int(
        len(identity_forensic)
    ),

    "full_response_overlap_files": int(
        full_response_count
    ),

    "full_session_overlap_files": int(
        full_session_count
    ),

    "partial_overlap_files": int(
        partial_overlap_count
    ),

    "zero_overlap_files": int(
        zero_overlap_count
    ),

    "canonical_population": {
        "responses": 35_072,
        "sessions": 22_821,
        "objectives": 398,
    },

    "test_population_locked": False,

    "production_inference_started": False,

    "submission_generated": False,

    "forensic_artifact": str(
        FORENSIC_PATH
    ),

    "next_action": (
        "If Dataset has no unseen identity candidate, "
        "search outside Dataset/ for the actual competition "
        "test/submission population before locking."
    ),
}


with open(
    SUMMARY_PATH,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        summary,
        f,
        indent=2,
    )


assert SUMMARY_PATH.exists()

print(
    f"Summary : {SUMMARY_PATH}"
)

print(
    "Summary write : PASS"
)


# =============================================================================
# 12. FINAL STATUS
# =============================================================================

print("\n" + "=" * 100)
print("TRACE THE ACE — CELL 3 FINAL STATUS")
print("=" * 100)

print(
    "Cell 2 dependency                 : PASS"
)

print(
    "Full-response overlap forensic    : PASS"
)

print(
    "Dataset population structure      : PASS"
)

print(
    "Canonical-size audit              : PASS"
)

print(
    "Unusual population audit          : PASS"
)

print(
    f"Zero-overlap candidates           : "
    f"{zero_overlap_count}"
)

print(
    f"Partial-overlap candidates        : "
    f"{partial_overlap_count}"
)

print(
    f"Dataset test evidence             : "
    f"{dataset_test_population_status}"
)

print(
    "Test population lock              : NOT YET"
)

print(
    "Production inference              : NOT STARTED"
)

print(
    "Submission generation             : NOT STARTED"
)

print("-" * 100)

print(
    "CELL 3 COMPLETE — PASS"
)

print("=" * 100)
```
